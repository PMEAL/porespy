# `ps_round`

Creates a circular (`ndim=2`) or spherical (`ndim=3`) structuring element of a given radius.

```python
import matplotlib.pyplot as plt

import porespy as ps

ps.visualization.set_mpl_style()
```

## `r` and `ndim`

Setting `ndim=2` returns a 2D disk (equivalent to `ps_disk`), while `ndim=3` returns a 3D ball (equivalent to `ps_ball`):

```python
s2d = ps.tools.ps_round(r=8, ndim=2)
s3d = ps.tools.ps_round(r=8, ndim=3)
print(f"ndim=2 shape: {s2d.shape}")
print(f"ndim=3 shape: {s3d.shape}")

fig, ax = plt.subplots(1, 2, figsize=[8, 4])
ax[0].imshow(s2d, origin="lower", interpolation="none")
ax[0].set_title("ndim=2 (r=8)")
ax[0].axis(False)
ax[1].imshow(ps.visualization.sem(s3d), cmap=plt.cm.bone)
ax[1].set_title("ndim=3 (r=8, SEM view)")
ax[1].axis(False);
```

## `smooth`

Controls whether the single-voxel nibs at exact distance `r` along the coordinate axes are excluded (`smooth=True`, default) or included (`smooth=False`):

```python
s_smooth = ps.tools.ps_round(r=6, ndim=2, smooth=True)
s_nibs = ps.tools.ps_round(r=6, ndim=2, smooth=False)

fig, ax = plt.subplots(1, 2, figsize=[8, 4])
ax[0].imshow(s_smooth, origin="lower", interpolation="none")
ax[0].set_title("smooth=True")
ax[0].axis(False)
ax[1].imshow(s_nibs, origin="lower", interpolation="none")
ax[1].set_title("smooth=False")
ax[1].axis(False);
```
