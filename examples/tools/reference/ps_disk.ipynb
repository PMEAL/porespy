# `ps_disk`

Creates a 2D circular disk structuring element for morphological operations.

```python
import matplotlib.pyplot as plt

import porespy as ps

ps.visualization.set_mpl_style()
```

## `r`

Specify the radius `r` of the circular disk structuring element:

```python
d5 = ps.tools.ps_disk(r=5)
d15 = ps.tools.ps_disk(r=15)
print(f"r=5 shape:  {d5.shape}")
print(f"r=15 shape: {d15.shape}")

fig, ax = plt.subplots(1, 2, figsize=[8, 4])
ax[0].imshow(d5, origin="lower", interpolation="none")
ax[0].set_title("r=5")
ax[0].axis(False)
ax[1].imshow(d15, origin="lower", interpolation="none")
ax[1].set_title("r=15")
ax[1].axis(False);
```

## `smooth`

When `smooth=True` (default), voxels at exact distance `r` are excluded, removing the single-voxel nibs on the four cardinal faces. Setting `smooth=False` includes those nibs (matching `skimage.morphology.disk`).

```python
d_smooth = ps.tools.ps_disk(r=6, smooth=True)
d_nibs = ps.tools.ps_disk(r=6, smooth=False)

fig, ax = plt.subplots(1, 2, figsize=[8, 4])
ax[0].imshow(d_smooth, origin="lower", interpolation="none")
ax[0].set_title("smooth=True")
ax[0].axis(False)
ax[1].imshow(d_nibs, origin="lower", interpolation="none")
ax[1].set_title("smooth=False")
ax[1].axis(False);
```
