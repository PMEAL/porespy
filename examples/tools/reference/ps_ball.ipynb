# `ps_ball`

Creates a 3D spherical ball structuring element for morphological operations.

```python
import matplotlib.pyplot as plt

import porespy as ps

ps.visualization.set_mpl_style()
```

## `r`

Specify the radius `r` of the 3D spherical structuring element:

```python
b5 = ps.tools.ps_ball(r=5)
b15 = ps.tools.ps_ball(r=15)
print(f"r=5 shape:  {b5.shape}")
print(f"r=15 shape: {b15.shape}")

fig, ax = plt.subplots(1, 2, figsize=[8, 4])
ax[0].imshow(ps.visualization.sem(b5), cmap=plt.cm.bone)
ax[0].set_title("r=5 (SEM view)")
ax[0].axis(False)
ax[1].imshow(ps.visualization.sem(b15), cmap=plt.cm.bone)
ax[1].set_title("r=15 (SEM view)")
ax[1].axis(False);
```

## `smooth`

When `smooth=True` (default), the single-voxel nibs on the six axial faces are omitted. Setting `smooth=False` retains them:

```python
b_smooth = ps.tools.ps_ball(r=6, smooth=True)
b_nibs = ps.tools.ps_ball(r=6, smooth=False)
mid = b_smooth.shape[2] // 2

fig, ax = plt.subplots(1, 2, figsize=[8, 4])
ax[0].imshow(b_smooth[:, :, mid], origin="lower", interpolation="none")
ax[0].set_title("smooth=True (mid-slice)")
ax[0].axis(False)
ax[1].imshow(b_nibs[:, :, mid], origin="lower", interpolation="none")
ax[1].set_title("smooth=False (mid-slice)")
ax[1].axis(False);
```
