# `ps_rect`

Creates a rectilinear (square in 2D, cubic in 3D) structuring element of width `w` for morphological operations.

```python
import matplotlib.pyplot as plt
import numpy as np

import porespy as ps

ps.visualization.set_mpl_style()
```

## `w` and `ndim`

Specify the side length `w` and dimensionality `ndim` (`2` for a square, `3` for a cube):

```python
r2d = ps.tools.ps_rect(w=7, ndim=2)
r3d = ps.tools.ps_rect(w=7, ndim=3)
print(f"ndim=2 shape: {r2d.shape}, dtype: {r2d.dtype}")
print(f"ndim=3 shape: {r3d.shape}, dtype: {r3d.dtype}")
```

Using `ps_rect` as the structuring element in a morphological erosion or dilation:

```python
im = ps.generators.blobs(shape=[150, 150], porosity=0.6, seed=0)
eroded = ps.filters.fftmorphology(im=im, strel=ps.tools.ps_rect(w=7, ndim=2), mode="erosion")

fig, ax = plt.subplots(1, 2, figsize=[10, 5])
ax[0].imshow(im, origin="lower", interpolation="none")
ax[0].set_title("Original image")
ax[0].axis(False)
ax[1].imshow(eroded, origin="lower", interpolation="none")
ax[1].set_title("Eroded with ps_rect(w=7, ndim=2)")
ax[1].axis(False);
```
